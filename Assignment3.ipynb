```python
import pandas as pd
```

```python
#5 create a dataframe
```

```python
data = {
    "Name" : ["Amy","Kajal","Diya","Thushita","Delna","Saujanya","Bargavi","Jane","Aswathy","Ryan Gosling"],
    "Age" : [30,22,23,21,24,26,89,19,34,48],
    "Gender" : ["Female","Female","Female","Female","Female","Female","Female","Female","Female","Male"]
}
df = pd.DataFrame(data)
print(df)
```

```python
#Add a new coloumn as "occupation"
```

```python
df["Occupation"]= ["Programmer", "Manager", "Analyst",
    "Programmer", "Manager", "Analyst",
    "Programmer", "Manager", "Analyst",
    "Programmer"]
print(df)
```

```python
#Select the rows of the dataframe where the age is greater than or equal to 30.
```

```python
ages_df = df[df["Age"] >= 30]
print(ages_df)
```
