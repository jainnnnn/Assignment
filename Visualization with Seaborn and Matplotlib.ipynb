```python
# import libraries
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# load titanic dataset
```

```python
titanic = sns.load_dataset("titanic")
```

```python
# 1.Countplot: Passenger Count by Embarked Port
```

```python
plt.figure(figsize=(8,6))
sns.countplot(data=titanic, x="embarked", hue="sex")

plt.title("Passenger Count by Embarked Port")
plt.xlabel("Embarked Port")
plt.ylabel("Number of Passengers")
plt.legend(title="sex")
plt.show()

print("Observations:")
print("- Most passengers in First Class are adults.")
print("- Third Class has a wider spread of ages, including many children.")
print("- Male and female passengers are present in all classes.\n")
```

```python
# 2. Swarmplot: Age Distribution across Classes
```

```python
plt.figure(figsize=(10,6))
sns.swarmplot(data=titanic, x="pclass", y="age", hue="sex", size=3)

plt.title("Age Distribution across Passenger Classes")
plt.xlabel("Passenger Class")
plt.ylabel("Age")
plt.legend(title="Sex")
plt.show()

print("Observation:")
print("- Most passengers in First Class are adults.")
print("- Third Class has a wider spread of ages, including many children.")
print("- Male and female passengers are present in all classes.\n")
```

```python
#3. Pie Chart: Gender Distribution
```

```python
gender_count = titanic["sex"].value_counts()

plt.figure(figsize=(6,6))
plt.pie(
    gender_count,
    labels=gender_count.index,
    autopct="%1.1f%%",
    explode=(0.00,0),
    startangle=90,
    shadow=True
)
plt.title("Gender Distribution")
plt.show()

print("Observation:")
print("- Male passengers form the majority of the dataset.")
print("- The explode effect highlights the male category.\n")
```

```python
# 4. Heatmap: Correlation Matrix
```

```python
columns = ["age", "fare", "pclass", "sibsp", "parch", "survived"]

correlation = titanic[columns].corr()

plt.figure(figsize=(8,6))
sns.heatmap(correlation, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Matrix")
plt.show()

print("Observation:")
print("- Fare and passenger class have a strong negative correlation.")
print("- Survival has a slight positive correlation with fare.")
print("- Most other correlations are relatively weak.\n")
```

```python
# 5. Violin Plot: Fare by Class and Gender
```

```python
plt.figure(figsize=(9,6))
sns.violinplot(
    data=titanic,
    x="pclass",
    y="fare",
    hue="sex",
    split=True
)

plt.title("Fare Distribution by Passenger Class and Gender")
plt.xlabel("Passenger Class")
plt.ylabel("Fare")
plt.legend(title="Sex")
plt.show()

print("Observation:")
print("- First Class passengers paid the highest fares.")
print("- Third Class fares are generally much lower.")
print("- Fare distributions are similar for males and females within each class.\n")
```

```python
# 6. Scatter Plot: Age vs Fare
```

```python
plt.figure(figsize=(9,6))
sns.scatterplot(
    data=titanic,
    x="age",
    y="fare",
    hue="survived"
)

plt.title("Age vs Fare by Survival Status")
plt.xlabel("Age")
plt.ylabel("Fare")
plt.legend(title="Survived")
plt.show()

print("Observation:")
print("- Higher fare passengers appear to have a higher survival rate.")
print("- Most passengers paid lower fares regardless of age.")
print("- There is no strong relationship between age and fare.")
```
