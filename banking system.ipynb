```python
#sample account data

accounts = {
    "1001" : {
        "name" : "JAIN",
        "pin" : "1234",
        "balance" : 500
        },
    "1002" : {
        "name" : "JAINY",
        "pin" : "4321",
        "balance" : 3999
        }
    }
    
```

```python
#Login function

def login():
  print("====== BANK LOGIN ======")

  account_no = input("Enter account number: ")
  pin = input("Enter PIN: ")

  if account_no in accounts :
     if accounts[account_no]["pin"] == pin:
      print("\nLogin successful!")
      print("Welcome,", accounts[account_no]["name"])
      return account_no
     else:
      print("Incorrect PIN!")
  else:
      print("Accounts does not exist!")
      return None
```

```python
#Deposit function

def deposit(account_no):
    amount = float(input("Enter the amount to deposit: ₹"))

    if amount > 0:
        accounts[account_no]["balance"] += amount
        print("Deposit Successful!")
        print("Current Balance:", accounts[account_no]["balance"])
    else:
        print("Deposit amount must be greater than zero.")
```

```python
#Withdrawal function

def withdraw(account_no):
    amount = float(input("Enter amount to withdraw: ₹"))

    if amount <= 0:
        print("Withdrawal amount must be greater than zero.")

    elif amount > accounts[account_no]["balance"]:
        print("Insufficient Balance!")

    else:
        accounts[account_no]["balance"] -= amount
        print("Withdrawal Successful!")
        print("Current Balance:", accounts[account_no]["balance"])
```

```python
#Balance function

def check_balance(account_no):
    print("Current Balance: ₹", accounts[account_no]["balance"])
```

```python

user = login()

if user:
    while True:
        print("\n====== BANK MENU ======")
        print("1. Deposit")
        print("2. Withdraw")
        print("3. Check Balance")
        print("4. Exit")

        choice = input("Enter your choice: ")

        if choice == "1":
            deposit(user)

        elif choice == "2":
            withdraw(user)

        elif choice == "3":
            check_balance(user)

        elif choice == "4":
            print("Thank you for using our Banking System.")
            break

        else:
            print("Invalid choice! Please try again!")
```
